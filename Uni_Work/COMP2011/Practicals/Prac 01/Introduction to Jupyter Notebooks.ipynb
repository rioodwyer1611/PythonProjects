# Introduction to Jupyter Notebook

Jupyter notebook or ipython notebook is an interactive computational environment where we can execute the code, add rich text, insert images as well as math equations. You can think it as a laboratory notebook for scientific computing where you not only mentions results but also enriches it with text such as experiment procedure and interpretation of results. 

In this practical, you will be introduced to using Jupyter Notebook. First, you will learn important concepts regarding the notebook. Then you will attempt the exercise to apply and test your knowledge.

## 2. Jupyter Notebook Interface

Let's understand the components of the notebook interface.

![notebook_components](images/notebook-components1.png)

<i><center><font size=2> Figure 3. Notebook interface components </font></center></i>

1. This is the name of the notebook. Currently it is "Untitled". Click on the name to change the name of the notebook.
2. This is the menubar. The important funcionalities from the menubar will be described as the notebook progresses.
3. This is the toolbar. It has important shortcuts for the functionalities from the menubar.
4. The large rectangle encompasses the cell. You can write code or add text/image in the cell. The cell are are mainly of two types - Code and Markdown. Code cells are useful to execute programm code whereas markdown cells are used for formatting i.e. adding text, images, etc.
5. It shows the kernel currently in use in the notebook. By default 'No Kernel' is selected. The circle next to the kernel name shows status of the kernel. When the circle is empty (as shown in the figure-3), it indicates the kernel is idle. If it's a bold circle, it means the kernel is busy indicating code is being executed in code cells.
